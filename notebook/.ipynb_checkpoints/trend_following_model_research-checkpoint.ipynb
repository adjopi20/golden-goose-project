```python
# Step 2 — Load your parquet

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)

df = pd.read_parquet("../research/080626-1535/1535.parquet")  # adjust path if needed

print("rows:", len(df))
print("columns:", len(df.columns))
df.head()
```

```python
# Step 3 — Convert timestamps

df["event_dt"] = pd.to_datetime(df["event_timestamp"], unit="ms", utc=True)
df["confirmation_dt"] = pd.to_datetime(df["setup_confirmation_timestamp"], unit="ms", utc=True)

df["hour_utc"] = df["event_dt"].dt.hour
df["minute_utc"] = df["event_dt"].dt.minute
df["minute_of_day_utc"] = df["hour_utc"] * 60 + df["minute_utc"]

df[["event_dt", "confirmation_dt", "directional_bias", "bubble_tier", "mfe_3600s_pct", "efficiency_3600s"]].head()
```

```python
# Step 4 — Basic distributions
    # MFE distribution

plt.figure(figsize=(10, 5))
plt.hist(df["mfe_3600s_pct"].dropna() * 100, bins=100)
plt.title("MFE 3600s Distribution")
plt.xlabel("MFE 3600s (%)")
plt.ylabel("Event Count")
plt.show()
```

```python
# Step 4 — Basic distributions
    # MAE distribution

plt.figure(figsize=(10, 5))
plt.hist(df["mae_3600s_pct"].dropna() * 100, bins=100)
plt.title("MAE 3600s Distribution")
plt.xlabel("MAE 3600s (%)")
plt.ylabel("Event Count")
plt.show()
```

```python
# Step 4 — Basic distributions
    # efficiency distribution

plt.figure(figsize=(10, 5))
plt.hist(df["efficiency_3600s"].dropna(), bins=100)
plt.title("Efficiency 3600s Distribution")
plt.xlabel("Efficiency")
plt.ylabel("Event Count")
plt.show()
```

```python
# Step 5 — Create outcome labels
    # Percentile-based labels

target_col = "mfe_3600s_pct"

q95 = df[target_col].quantile(0.95)
q99 = df[target_col].quantile(0.99)
q995 = df[target_col].quantile(0.995)
q999 = df[target_col].quantile(0.999)

print("95% cutoff:", q95, "=", q95 * 100, "%")
print("99% cutoff:", q99, "=", q99 * 100, "%")
print("99.5% cutoff:", q995, "=", q995 * 100, "%")
print("99.9% cutoff:", q999, "=", q999 * 100, "%")

df["top_5pct_mfe"] = df[target_col] >= q95
df["top_1pct_mfe"] = df[target_col] >= q99
df["top_05pct_mfe"] = df[target_col] >= q995
df["top_01pct_mfe"] = df[target_col] >= q999

df[["top_5pct_mfe", "top_1pct_mfe", "top_05pct_mfe", "top_01pct_mfe"]].sum()
```

```python
# Step 5 — Create outcome labels
    # Clean good event label This is better than MFE alone:

df["clean_good_3600"] = (
    (df["mfe_3600s_pct"] >= 0.005) &       # 0.50% favorable move
    (df["efficiency_3600s"] >= 0.60) &     # clean directional movement
    (df["mae_3600s_pct"] <= 0.0035)        # max 0.35% adverse move
)

df["clean_good_3600"].value_counts(normalize=True)
```

```python
# Step 5 — Create outcome labels
    # Clean good event label This is better than MFE alone:

df["runner_3600"] = (
    (df["mfe_3600s_pct"] >= 0.01) &        # 1.00% favorable move
    (df["efficiency_3600s"] >= 0.60) &
    (df["mae_3600s_pct"] <= 0.005)
)

df["runner_3600"].value_counts(normalize=True)
```

```python
# Step 6 — Compare good vs bad events

compare_cols = [
    "mfe_3600s_pct",
    "mae_3600s_pct",
    "efficiency_3600s",
    "reaction_mfe_30s_pct",
    "reaction_mae_30s_pct",
    "reaction_efficiency_30s",
    "distance_from_poc_pct",
    "distance_from_vah_pct",
    "distance_from_val_pct",
    "anchor_bubble_qty",
    "anchor_bubble_notional",
]

df.groupby("clean_good_3600")[compare_cols].median().T
```

```python
# Step 6 — Compare good vs bad events
    # Also count by direction:

pd.crosstab(df["directional_bias"], df["clean_good_3600"], normalize="index")
```

```python
# Step 6 — Compare good vs bad events
    # By bubble tier:

pd.crosstab(df["bubble_tier"], df["clean_good_3600"], normalize="index")
```

```python
# Step 6 — Compare good vs bad events
    # By location:

pd.crosstab(df["location"], df["clean_good_3600"], normalize="index")
```

```python
# Step 7 — Create feature buckets

    # Reaction efficiency bucket
df["reaction_eff_bucket"] = pd.cut(
    df["reaction_efficiency_30s"],
    bins=[0, 0.80, 0.90, 0.95, 0.98, 1.00],
    labels=["0.80-0.90", "0.90-0.95", "0.95-0.98", "0.98-1.00", "1.00"],
    include_lowest=True
)

    # Distance from POC bucket
df["abs_distance_from_poc_pct"] = df["distance_from_poc_pct"].abs()

df["poc_distance_bucket"] = pd.qcut(
    df["abs_distance_from_poc_pct"],
    q=5,
    duplicates="drop"
)

    # Time bucket
def time_bucket_utc(minute):
    # adjust these later based on your session definitions
    if 0 <= minute < 7 * 60:
        return "Asia"
    elif 7 * 60 <= minute < 12 * 60:
        return "Europe"
    elif 12 * 60 <= minute < 17 * 60:
        return "NY_AM"
    elif 17 * 60 <= minute < 21 * 60:
        return "NY_PM"
    else:
        return "Late_US"

df["time_bucket_utc"] = df["minute_of_day_utc"].apply(time_bucket_utc)

df["time_bucket_utc"].value_counts()
```

```python
# Step 7 — Create feature buckets
    # Since your session starts at 13:30 UTC, you can also create minutes since session open:

session_start_minute = 13 * 60 + 30

df["minutes_since_session_open"] = (
    df["minute_of_day_utc"] - session_start_minute
) % (24 * 60)

df["session_phase"] = pd.cut(
    df["minutes_since_session_open"],
    bins=[0, 60, 180, 360, 720, 1080, 1440],
    labels=["0-1h", "1-3h", "3-6h", "6-12h", "12-18h", "18-24h"],
    include_lowest=True
)
```

```python
# Step 8 — Conditional probability tables
    # By direction + tier

summary = (
    df.groupby(["directional_bias", "bubble_tier"], observed=True)
      .agg(
          events=("event_id", "count"),
          good_rate=("clean_good_3600", "mean"),
          runner_rate=("runner_3600", "mean"),
          median_mfe=("mfe_3600s_pct", "median"),
          median_mae=("mae_3600s_pct", "median"),
          median_eff=("efficiency_3600s", "median"),
      )
      .reset_index()
      .sort_values(["good_rate", "events"], ascending=[False, False])
)

summary
```

```python
# Step 8 — Conditional probability tables
    # By time bucket

time_summary = (
    df.groupby(["time_bucket_utc", "directional_bias"], observed=True)
      .agg(
          events=("event_id", "count"),
          good_rate=("clean_good_3600", "mean"),
          runner_rate=("runner_3600", "mean"),
          median_mfe=("mfe_3600s_pct", "median"),
          median_eff=("efficiency_3600s", "median"),
      )
      .reset_index()
      .sort_values("good_rate", ascending=False)
)

time_summary
```

```python
# Step 8 — Conditional probability tables
    # By POC distance

poc_summary = (
    df.groupby(["directional_bias", "poc_distance_bucket"], observed=True)
      .agg(
          events=("event_id", "count"),
          good_rate=("clean_good_3600", "mean"),
          runner_rate=("runner_3600", "mean"),
          median_mfe=("mfe_3600s_pct", "median"),
          median_mae=("mae_3600s_pct", "median"),
          median_eff=("efficiency_3600s", "median"),
      )
      .reset_index()
      .sort_values("good_rate", ascending=False)
)

poc_summary
```

```python
# Step 8 — Conditional probability tables
    # Multi-condition table

combo_summary = (
    df.groupby(
        ["directional_bias", "time_bucket_utc", "bubble_tier", "reaction_eff_bucket"],
        observed=True
    )
    .agg(
        events=("event_id", "count"),
        good_rate=("clean_good_3600", "mean"),
        runner_rate=("runner_3600", "mean"),
        median_mfe=("mfe_3600s_pct", "median"),
        median_mae=("mae_3600s_pct", "median"),
        median_eff=("efficiency_3600s", "median"),
    )
    .reset_index()
)

# avoid tiny-sample traps
combo_summary = combo_summary[combo_summary["events"] >= 50]

combo_summary.sort_values("good_rate", ascending=False).head(30)
```

```python
# 9. Histogram by Groups
    # MFE by direction
for direction in df["directional_bias"].dropna().unique():
    subset = df[df["directional_bias"] == direction]
    plt.figure(figsize=(10, 5))
    plt.hist(subset["mfe_3600s_pct"].dropna() * 100, bins=100)
    plt.title(f"MFE 3600s Distribution - {direction}")
    plt.xlabel("MFE 3600s (%)")
    plt.ylabel("Event Count")
    plt.show()
```

```python
# 9. Histogram by Groups
    # Efficiency by good/bad event

plt.figure(figsize=(10, 5))
plt.hist(df.loc[~df["clean_good_3600"], "efficiency_3600s"].dropna(), bins=80, alpha=0.6, label="Not good")
plt.hist(df.loc[df["clean_good_3600"], "efficiency_3600s"].dropna(), bins=80, alpha=0.6, label="Good")
plt.title("Efficiency Distribution: Good vs Not Good")
plt.xlabel("Efficiency 3600s")
plt.ylabel("Event Count")
plt.legend()
plt.show()
```

```python
# 9. Histogram by Groups
    # Reaction MFE distribution

plt.figure(figsize=(10, 5))
plt.hist(df.loc[~df["clean_good_3600"], "reaction_mfe_30s_pct"].dropna() * 100, bins=80, alpha=0.6, label="Not good")
plt.hist(df.loc[df["clean_good_3600"], "reaction_mfe_30s_pct"].dropna() * 100, bins=80, alpha=0.6, label="Good")
plt.title("Reaction MFE 30s: Good vs Not Good")
plt.xlabel("Reaction MFE 30s (%)")
plt.ylabel("Event Count")
plt.legend()
plt.show()
```

```python
# Step 10 — Group overlapping good events into impulse groups
    # Start with clean_good_3600

good = df[df["clean_good_3600"]].copy()

good = good.sort_values(["session_id", "directional_bias", "event_timestamp"])

group_gap_seconds = 900  # 15 minutes; tune this later

good["gap_seconds"] = (
    good.groupby(["session_id", "directional_bias"])["event_timestamp"]
        .diff()
        .div(1000)
)

good["new_group"] = good["gap_seconds"].isna() | (good["gap_seconds"] > group_gap_seconds)

good["impulse_group_number"] = (
    good.groupby(["session_id", "directional_bias"])["new_group"]
        .cumsum()
)

good["impulse_group_key"] = (
    good["session_id"].astype(str)
    + "_"
    + good["directional_bias"].astype(str)
    + "_"
    + good["impulse_group_number"].astype(str)
)

print("good overlapping events:", len(good))
print("impulse groups:", good["impulse_group_key"].nunique())
```

```python
# Step 10 — Group overlapping good events into impulse groups
    # Inspect group sizes:

group_sizes = good.groupby("impulse_group_key").size().sort_values(ascending=False)

group_sizes.describe(), group_sizes.head(20)
```

```python
# Step 10 — Group overlapping good events into impulse groups
    # Histogram of group sizes:

plt.figure(figsize=(10, 5))
plt.hist(group_sizes, bins=50)
plt.title("Good Event Impulse Group Sizes")
plt.xlabel("Events per impulse group")
plt.ylabel("Group count")
plt.show()
```

```python
# Step 11 — Choose one representative event per group
    # For validation, choose the earliest qualifying event.

representatives = (
    good.sort_values("event_timestamp")
        .groupby("impulse_group_key", as_index=False)
        .first()
)

print("representative events:", len(representatives))

representatives[
    ["impulse_group_key", "event_dt", "directional_bias", "bubble_tier", "mfe_3600s_pct", "efficiency_3600s"]
].head()
```

```python
# Step 12 — Compare overlapping good events vs representatives
    # by category

cols = [
    "mfe_3600s_pct",
    "mae_3600s_pct",
    "efficiency_3600s",
    "reaction_mfe_30s_pct",
    "reaction_efficiency_30s",
    "distance_from_poc_pct",
    "anchor_bubble_qty",
    "anchor_bubble_notional",
]

comparison = pd.DataFrame({
    "all_good_median": good[cols].median(),
    "representative_median": representatives[cols].median(),
})

comparison
```

```python
# Step 12 — Compare overlapping good events vs representatives
    # By direction:

rep_summary = (
    representatives.groupby(["directional_bias", "bubble_tier"], observed=True)
    .agg(
        groups=("impulse_group_key", "count"),
        median_mfe=("mfe_3600s_pct", "median"),
        median_mae=("mae_3600s_pct", "median"),
        median_eff=("efficiency_3600s", "median"),
        median_reaction_mfe=("reaction_mfe_30s_pct", "median"),
        median_reaction_eff=("reaction_efficiency_30s", "median"),
    )
    .reset_index()
    .sort_values("groups", ascending=False)
)

rep_summary
```

```python
# Step 13 — Find what good representatives have in common
    # Now compare representatives against all events.

df["is_representative_good"] = df["event_id"].isin(representatives["event_id"])

feature_cols = [
    "mfe_3600s_pct",
    "mae_3600s_pct",
    "efficiency_3600s",
    "reaction_mfe_30s_pct",
    "reaction_mae_30s_pct",
    "reaction_efficiency_30s",
    "distance_from_poc_pct",
    "distance_from_vah_pct",
    "distance_from_val_pct",
    "anchor_bubble_qty",
    "anchor_bubble_notional",
    "bubble_percentile_score",
]

df.groupby("is_representative_good")[feature_cols].median().T
```

```python
# Step 13 — Find what good representatives have in common
    # Categorical comparison:

for col in ["directional_bias", "location", "bubble_tier", "time_bucket_utc", "session_phase"]:
    print("\n", col)
    display(pd.crosstab(df[col], df["is_representative_good"], normalize="index"))
```

```python
# Step 14 — Base-rate adjusted probability
    # eg. P(good event | NY)

prob_table = (
    df.groupby(["time_bucket_utc", "directional_bias"], observed=True)
    .agg(
        total_events=("event_id", "count"),
        representative_good_rate=("is_representative_good", "mean"),
        clean_good_rate=("clean_good_3600", "mean"),
        runner_rate=("runner_3600", "mean"),
    )
    .reset_index()
)

prob_table.sort_values("representative_good_rate", ascending=False)
```

```python
# Step 15 — Find top candidate condition combinations

condition_summary = (
    df.groupby(
        ["directional_bias", "time_bucket_utc", "session_phase", "bubble_tier", "reaction_eff_bucket"],
        observed=True
    )
    .agg(
        events=("event_id", "count"),
        clean_good_rate=("clean_good_3600", "mean"),
        runner_rate=("runner_3600", "mean"),
        median_mfe=("mfe_3600s_pct", "median"),
        median_mae=("mae_3600s_pct", "median"),
        median_eff=("efficiency_3600s", "median"),
    )
    .reset_index()
)

condition_summary = condition_summary[condition_summary["events"] >= 30]

condition_summary.sort_values(
    ["clean_good_rate", "events"],
    ascending=[False, False]
).head(50)
```

```python
# Step 16 — Save your analysis outputs

output_dir = "../research/analysis_outputs"

import os
import pandas as pd

os.makedirs(output_dir, exist_ok=True)

summary.to_csv(f"{output_dir}/summary_direction_tier.csv", index=False)
time_summary.to_csv(f"{output_dir}/summary_time.csv", index=False)
combo_summary.to_csv(f"{output_dir}/summary_combo.csv", index=False)
condition_summary.to_csv(f"{output_dir}/condition_summary.csv", index=False)

representatives_export = representatives.copy()

for col in representatives_export.columns:
    dtype = representatives_export[col].dtype

    if isinstance(dtype, pd.CategoricalDtype) or isinstance(dtype, pd.IntervalDtype):
        representatives_export[col] = representatives_export[col].astype(str)

representatives_export.to_parquet(
    f"{output_dir}/representative_good_events.parquet",
    index=False
)
```
